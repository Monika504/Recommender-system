```python
# Load the TMDB movie metadata and credits datasets into Pandas DataFrames for analysis.
movies = pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_movies.csv')
credits = pd.read_csv('/kaggle/input/tmdb-movie-metadata/tmdb_5000_credits.csv') 
```

```python
# Merges the movies and credits DataFrames on the title column to combine movie details with corresponding credits.
movies = movies.merge(credits,on='title')
```

```python
# after initial data exploration, following are the columns that have most impact on recommendation,
# hence dropping other columns
movies = movies[['movie_id','title','overview','genres','keywords','cast','crew']]
```

**Data Preprocessing**

Now we will combine the terms in overview, genres,keywords, cast columns into a single column using string manupulation methods



```python
movies.info()
```

```python
# dropping the null values
movies.dropna(inplace=True)
movies.isna().sum()
```

```python
movies.iloc[0].genres
```

```python
movies['genres'].dtype
```

```python
# Parse the string representation of movie's genres into a Python list of dictionaries using ast.literal_eval.
import ast
ast.literal_eval(movies['genres'].iloc[0])
```

```python
# function to extract genre names into a list
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L 
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
movies['cast'] = movies['cast'].apply(convert)
movies.head()
```

```python
# slice the cast names to keep only the names of 3 lead actors
movies['cast'] = movies['cast'].apply(lambda x:x[0:3])
```

```python
# function to extract teh director's name from the cast column
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
movies.sample(5)
```

```python
# in order to remove whitespaces so as to avoid confusion between different people with same name

def collapse(L):
    return [i.replace(" ", "") for i in L]
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
movies.head()
```

```python
# Split each movie overview into a list of words using whitespace as the delimiter.
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
# Create a new column 'tags' by concatenating the 'overview', 'genres', 'keywords', 'cast', and 'crew' columns for each movie.
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
# Drops the columns 'overview', 'genres', 'keywords', 'cast', and 'crew' from the movies DataFrame, keeping only the remaining relevant data.
new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
# Convert the list of words in the 'tags' column into a single string by joining the words with a space
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

```python
new.iloc[0].tags
```

```python
# vectorization of each tag
# bag of words techinique

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=8000,stop_words='english')
    
vector = cv.fit_transform(new['tags']).toarray()
vector.shape
```

```python
vector[0]
```

```python
# Return the list of the feature names (vocabulary) generated by the CountVectorizer (cv) after fitting it to the data.
cv.get_feature_names_out().tolist()
```

```python
# Import the nltk library and initialize the PorterStemmer object (ps) for stemming words to their root form using the Porter stemming algorithm.
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
stem(new.iloc[0].tags)
```

```python
new['tags'] = new['tags'].apply(stem)
```

```python

vector = cv.fit_transform(new['tags']).toarray()
vector.shape
```

```python
cv.get_feature_names_out().tolist()
```

```python
# Calculate the cosine similarity between the movie vectors, which measures the similarity between the movies based on their 'tags', resulting in a similarity matrix.
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vector)
similarity
```

```python
# before arranging the similarity from highest to lowest, we have to retain the information of similarity to each movie 
list(enumerate(similarity[0]))

```

```python
sorted_similarity = sorted(list(enumerate(similarity[0])), key=lambda x: x[1], reverse=True)
print(sorted_similarity)
```

```python
def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
        
```

```python
recommend('Shutter Island')
```

```python
# Save the processed movie data (new) and the similarity matrix (similarity) into pickle files for future use or model deployment.
import pickle
pickle.dump(new,open('movie_list.pkl','wb'))
pickle.dump(similarity,open('similarity.pkl','wb'))
```
